```python

# ==========================================
# Xbox2Android - Hari 2
# Sistem Pemeriksaan File dan Metadata
# ==========================================

import json
import zipfile
from pathlib import Path

# 1. MENYIAPKAN FOLDER PROYEK

project_dir = Path("/content/Xbox2Android")
sample_dir = project_dir / "games" / "contoh"
output_dir = project_dir / "output"

sample_dir.mkdir(parents=True, exist_ok=True)
output_dir.mkdir(parents=True, exist_ok=True)

# 2. MEMBACA CONFIG.JSON

config_file = project_dir / "config.json"

if not config_file.exists():
    config_awal = {
        "nama": "Xbox2Android",
        "versi": "0.1",
        "target_konsol": "Xbox 360",
        "target_game": "Sonic Unleashed",
        "status": "Development"
    }

    with open(config_file, "w") as f:
        json.dump(config_awal, f, indent=4)

with open(config_file, "r") as f:
    config = json.load(f)

print("=== INFORMASI PROYEK ===")

for kunci, nilai in config.items():
    print(kunci, ":", nilai)

# 3. MEMBUAT FILE LATIHAN

fake_xex = sample_dir / "contoh.xex"
fake_xex.write_bytes(b"XEX2" + b"\x00" * 60)

fake_iso = sample_dir / "contoh.iso"
fake_iso.write_bytes(b"INI HANYA FILE ISO LATIHAN" * 10)

fake_zip = sample_dir / "contoh.zip"

with zipfile.ZipFile(fake_zip, "w") as archive:
    archive.writestr(
        "catatan.txt",
        "Ini hanya file latihan."
    )

print("\nFile latihan berhasil dibuat!")

# 4. FUNGSI PEMERIKSAAN FILE

def periksa_file_game(lokasi_file):

    file = Path(lokasi_file)

    hasil = {
        "nama": file.name,
        "ekstensi": file.suffix.lower(),
        "ukuran_byte": None,
        "status": "Belum diperiksa"
    }

    if not file.exists() or not file.is_file():
        hasil["status"] = "File tidak ditemukan"
        return hasil

    ukuran = file.stat().st_size

    hasil["ukuran_byte"] = ukuran
    hasil["ukuran_kb"] = round(ukuran / 1024, 2)

    ekstensi = file.suffix.lower()

    if ekstensi == ".xex":

        with open(file, "rb") as f:
            header = f.read(4)

        if header == b"XEX2":
            hasil["status"] = "Header XEX2 ditemukan"
        else:
            hasil["status"] = "Header tidak cocok"

    elif ekstensi == ".zip":

        if zipfile.is_zipfile(file):

            with zipfile.ZipFile(file, "r") as archive:
                hasil["isi_arsip"] = archive.namelist()

            hasil["status"] = "ZIP dapat dibaca"

        else:
            hasil["status"] = "ZIP tidak valid"

    elif ekstensi == ".iso":

        hasil["status"] = "Perlu pemeriksaan struktur ISO"

    else:

        hasil["status"] = "Format belum didukung"

    return hasil


# 5. MEMERIKSA SEMUA FILE

laporan = []

print("\n=== HASIL PEMERIKSAAN ===")

for file in sample_dir.iterdir():

    if file.is_file():

        hasil = periksa_file_game(file)
        laporan.append(hasil)

        print("\nNama :", hasil["nama"])
        print("Ukuran:", hasil["ukuran_byte"], "byte")
        print("Status:", hasil["status"])


# 6. MENYIMPAN LAPORAN JSON

report_file = output_dir / "laporan_hari2.json"

with open(report_file, "w") as f:
    json.dump(laporan, f, indent=4)

print("\n==============================")
print("LAPORAN BERHASIL DISIMPAN!")
print("Lokasi:", report_file)
print("==============================")

# 7. MENAMPILKAN ISI LAPORAN

print("\n=== ISI LAPORAN JSON ===")
print(json.dumps(laporan, indent=4))
```
